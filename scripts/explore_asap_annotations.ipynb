#### Try with Tim's chunking

```python
tim_asap_path = Path("~/Documents/transcription/MIDI2ScoreTransformer/midi2scoretransformer/data/asap-dataset").expanduser()
```

```python
tim_sample_path = tim_asap_path / sample_metadata["midi_performance"]
chunks = pd.read_json(tim_sample_path.parent / f"{tim_sample_path.stem}_chunks.json")
# Merge with beat_type sequentially, with NaN at beginning if not same length
num_nan = len(chunks) - len(sample_annotations)
beat_type = pd.concat([pd.Series([""] * num_nan), sample_annotations["beat_type"].copy(deep=True)], axis=0, ignore_index=True)
assert len(chunks) == len(beat_type)
chunks["beat_type"] = beat_type
chunks
```

```python
# Assign an index to all db groups
chunks["measure"] = chunks["beat_type"].str.startswith("db").cumsum()
# Aggregate by group
chunks_measure = chunks[["midi", "measure"]].groupby("measure").agg(
    func=lambda x: sum(x, []),
)
# Make sure all numbers are consecutive in "midi" when they are flattened
flatten = sum(chunks_measure["midi"], [])
assert all(flatten[i] == flatten[i - 1] + 1 for i in range(1, len(flatten)))
# Keep only the first element on each measure
chunks_measure["midi"] = chunks_measure["midi"].apply(lambda x: x[0])
chunks_measure
```

```python
# Keep only notes from the first to the last measure
clipped = performance.copy(deep=True)
clipped.tracks[0].notes = clipped.tracks[0].notes[chunks_measure.iloc[first_measure]["midi"]:chunks_measure.iloc[last_measure]["midi"]]
# Clip the performance to the timing of the first and last notes
start_time = clipped.tracks[0].notes[0].time
end_time = max(note.time + note.duration for note in clipped.tracks[0].notes)
clipped.clip(start_time, end_time, inplace=True)
# Adjust the timings so the performances starts at 0
clipped.adjust_time(original_times=[start_time, end_time], new_times=[0, end_time - start_time], inplace=True)
# Dump midi
clipped.dump_midi(performance_path.parent / f"{performance_path.stem}_M{first_measure}-{last_measure}_tim.mid")
```

```python
def clip_performance(performance: symusic.Score, chunks: pd.DataFrame, first_measure: int = None, last_measure: int = None):
    # Assign an index to all db groups
    chunks["measure"] = chunks["beat_type"].str.startswith("db").cumsum()
    # Aggregate by group
    chunks_measures: pd.DataFrame = chunks[["midi", "measure"]].groupby("measure").agg(
        func=lambda x: sum(x, []),
    )
    # Make sure all numbers are consecutive in "midi" when they are flattened
    flatten = sum(chunks_measures["midi"], [])
    assert all(flatten[i] == flatten[i - 1] + 1 for i in range(1, len(flatten)))
    # Keep only the first element on each measure
    chunks_measures["midi"] = chunks_measures["midi"].apply(lambda x: x[0])
    # Add last measure to the end
    chunks_measures.loc[len(chunks_measures)] = [len(performance.tracks[0].notes)]
    print(chunks_measures)
    # Setup default values
    if first_measure is None:
        first_measure = 0
    if last_measure is None:
        last_measure = len(chunks_measures) - 1
    # Keep only notes from the first to the last measure
    clipped = performance.copy(deep=True)
    clipped.tracks[0].notes = clipped.tracks[0].notes[chunks_measures.iloc[first_measure]["midi"]:chunks_measures.iloc[last_measure]["midi"]]
    # Clip the performance to the timing of the first and last notes
    start_time = clipped.tracks[0].notes[0].time
    end_time = max(note.time + note.duration for note in clipped.tracks[0].notes)
    clipped.clip(start_time, end_time, inplace=True)
    # Adjust the timings so the performances starts at 0
    clipped.adjust_time(original_times=[start_time, end_time], new_times=[0, end_time - start_time], inplace=True)
    return clipped

s, e = 17, 18
clipped = clip_performance(performance, chunks, s, e)
clipped.dump_midi(performance_path.parent / f"{performance_path.stem}_M{s}-{e}_tim.mid")
```
